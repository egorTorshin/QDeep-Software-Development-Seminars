### Variables, Data Types

```python
# Each object has three main properties: type, identity, value and methods
x = 5

print(type(x))
print(id(x))
print(x)
print(dir(x))
```

### Mutability vs Immutability

```python
# List is a mutable object
lst = [1, 2, 3, 4, 5]
print(id(lst))

lst.append(6)
print(id(lst))
```

```python
# Integer is an immutable object
x = 5
print(id(x))

x += 1
print(id(x))
```
