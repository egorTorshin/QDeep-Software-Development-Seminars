### Multithreading

```python
import threading
import time

def worker(name):
    for i in range(3):
        print(name, i)
        time.sleep(0.5)

t1 = threading.Thread(target=worker, args=("A",))
t2 = threading.Thread(target=worker, args=("B",))

t1.start()
t2.start()

t1.join()
t2.join()
```

### Async Programming

```python
import asyncio
import time


async def worker(name, delay):
    print(f"{time.strftime('%X')} start {name}")
    await asyncio.sleep(delay)
    print(f"{time.strftime('%X')} end   {name}")


async def main():
    task1 = asyncio.create_task(worker("A", 2))
    task2 = asyncio.create_task(worker("B", 2))

    await task1
    await task2


await main()
```
